### Why Detection (YOLO) Fails vs Density Regression
1. **Severe Occlusion**: In crowds with >1000 people, heads overlap by >80%, causing Non-Maximum Suppression (NMS) to eliminate valid predictions.
2. **Scale Distortion**: Perspective transforms make background heads 3-5 pixels wide while foreground heads are 40+ pixels wide.
3. **Density Map Solution**: Predicting a continuous density map $D(x, y)$ whose spatial integral $\iint D(x, y)\,dx\,dy = \text{Total People}$ bypasses bounding box localization.
