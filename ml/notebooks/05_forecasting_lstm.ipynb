# CrowdSense ML - Notebook 05: Crowd Count & Influx Forecasting (LSTM)

Predicts future crowd levels 5 to 30 minutes in advance using sequence-to-sequence LSTMs.
Enables proactive venue management before emergency capacity thresholds are breached.

```python
import numpy as np
import torch
import torch.nn as nn
import matplotlib.pyplot as plt

# Sample sequence forecast simulation
time_steps = np.arange(0, 60)
historical_counts = 150 + 50 * np.sin(time_steps / 10) + np.random.normal(0, 5, len(time_steps))
forecast_steps = np.arange(60, 75)
predicted_counts = 150 + 50 * np.sin(forecast_steps / 10) + 15

plt.figure(figsize=(9, 4))
plt.plot(time_steps, historical_counts, label='Observed Counts (FastAPI Ingestion)', color='#3478e5')
plt.plot(forecast_steps, predicted_counts, '--', label='LSTM Multi-step Forecast (+15 min)', color='#db9225')
plt.axhline(y=210, color='red', linestyle=':', label='Zone Critical Threshold (210 people)')
plt.title('Crowd Forecast & Proactive Bottleneck Warning')
plt.xlabel('Timeline (Minutes)')
plt.ylabel('People in Zone')
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
